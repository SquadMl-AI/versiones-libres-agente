```python
from ai_services import AzureServices
import pandas as pd
import os 
from dotenv import load_dotenv, find_dotenv
# Cargar variables de entorno (asegúrate que el .env tenga las de Azure)
load_dotenv(find_dotenv())
```

```python
db= os.getenv("AZURE_COSMOSDB_DATABASE_NAME")    
conection_tring= os.getenv("AZURE_COSMOSDB_ENDPOINT")
cosmos= AzureServices().CosmosDB(db_name=db, collection_names="Users_Auth", connection_string=conection_tring)
```

### Sentencias

```python
# Cargamos la lista de correos 
sentencias= pd.read_excel("./users/Listas Accesos IA-Sentencias CORREOS.xlsx") 
# Tomamos solo la columna de correos
sentencias= sentencias[["CORREO"]]
# Limpiamos los correos
sentencias["CORREO"]= sentencias["CORREO"].apply(lambda x: str(x).strip().lower().replace(" ", ""))
# Eliminamos duplicados
sentencias= sentencias[["CORREO"]].drop_duplicates()
# Añadimos la columna de módulo con el nombre del módulo al que el usuario tiene acceso
sentencias["MODULO"]= "sentencias"
```

```python
# Creamos un diccionario a partir del DataFrame de sentencias
diccionario_sentencias= sentencias.to_dict(orient= "records")
```

```python
# Guardar en cosmos
doc_sentencias= {
    "_id": "sentencias_usuarios",
    "users": diccionario_sentencias
    }

# Guardamos el documento en Cosmos DB
cosmos.update_users_field(doc=doc_sentencias,collection_name="Users_Auth")
```

### Audiencias

```python
audiencias= pd.read_excel("./users/Listas Accesos IA-Versiones CORREOS.xlsx") # <--- Vambiar tuta de ser necesario
audiencias= audiencias[["CORREO"]]
audiencias["CORREO"] = audiencias["CORREO"].apply(lambda x: str(x).strip().lower().replace(" ", ""))
audiencias = audiencias[["CORREO"]].drop_duplicates()
audiencias["MODULO"]= "audiencias"
```

```python
diccionario_audiencias= audiencias.to_dict(orient= "records")
```

### Insertar en Cosmos
