```python
import re
doc_name = "SENTENCIA-FERNEY2025ARGUMEDO.pdf"
def _extract_year_from_docname(doc_name: str):
    """
    Extrae el año más reciente de un nombre de documento buscando cualquier año de los últimos 100.
    """
    if not doc_name:
        return None

    # 1. Genera una lista de años para buscar, del más reciente al más antiguo.
    current_year = datetime.now().year
    years_to_check = range(current_year, current_year - 100, -1)

    # 2. Itera y devuelve la primera coincidencia (que será la más reciente).
    for year in years_to_check:
        if str(year) in doc_name:
            return year

    # 3. Si el bucle termina sin encontrar un año, devuelve None.
    return None

_extract_year_from_docname(doc_name)
```

```python
from azure.core.credentials import AzureKeyCredential
from azure.search.documents import SearchClient
from bs4 import BeautifulSoup
import re
import unicodedata
from azure.search.documents.models import (
    QueryType,
    VectorizedQuery,
    VectorQuery,
)

#Instanciamiento de los utils
blob_storage = AzureServices.AzureBlobStorage()
di = AzureServices.DocumentIntelligence()
aoai = AzureServices.AzureOpenAI()
search_client = AzureServices.AzureIASearch()

def normalize_text(query: str) -> str:
    """
    Normaliza la query para mejorar la búsqueda y la generación de respuestas:
    - Pasa a minúsculas
    - Elimina tildes (pero mantiene la ñ)
    - Elimina signos de puntuación y html
    """
    soup = BeautifulSoup(query, "html.parser")
    texto = soup.get_text(separator=" ").lower()
    # Reemplaza solo las vocales acentuadas, sin tocar la ñ
    reemplazos = {
        "á": "a", "é": "e", "í": "i", "ó": "o", "ú": "u",
        "ä": "a", "ë": "e", "ï": "i", "ö": "o", "ü": "u",
        "à": "a", "è": "e", "ì": "i", "ò": "o", "ù": "u"
    }
    for acentuada, simple in reemplazos.items():
        texto = texto.replace(acentuada, simple)
    # Elimina signos de puntuación
    texto = re.sub(r'[^\w\s]', '', texto)
    texto = re.sub(r'\s+', ' ', texto).strip()
    return texto

def hybrid_search(query: str, index_name: str) -> list[str]:
    """
    Realiza una búsqueda híbrida (vector + texto) con reclasificación semántica.

    Args:
        query (str): La consulta del usuario.
        index_name (str): Indice de Azure AI Search para buscar

    Returns:
        list[str]: Una lista de los fragmentos de contenido más relevantes.
    """
    # Este cliente se usa para buscar documentos en el indice definido por "index_name"
    search = SearchClient(
        endpoint=search_client.endpoint,
        index_name=index_name,
        credential=AzureKeyCredential(search_client.api_key)
    )
    # 1. Generar el vector para la consulta del usuario
    query_vector = aoai.get_embedding(query)

    # 2. Construir la consulta vectorial
    vector_query = VectorizedQuery(
        vector=query_vector, 
        k_nearest_neighbors=50, 
        fields="embedded_content_ltks"
    )

    print(f"🔍 Realizando búsqueda híbrida para: '{query}'")
    filter_string = "search.in(bloque, 'Bloque Calima', ',')"
    # 3. Ejecutar la búsqueda
    results = search.search(
        search_text=query,
        vector_queries=[vector_query],
        filter = filter_string,
        query_type=QueryType.SEMANTIC,  # Activa la reclasificación semántica
        semantic_configuration_name='semantic-config', # configuración semantica instanciada en AI Search
        top=50, # Número de resultados a devolver después de la reclasificación
        select=["doc_id", "content","content_ltks", "page_number", "bloque"],
        highlight_fields="content" # Resaltar los fragmentos de texto relevantes
    )

    thresholds = [2, 1.5, 1]
    top_content = []

    for threshold in thresholds:
        top_content = [content for content in results if content.get('@search.reranker_score', 0) >= threshold]
        if top_content:
            print(f"✅ Se encontraron {len(top_content)} resultados con score >= {threshold}")
            break
        else:
            print(f"🔎 No se encontraron resultados con umbral semantic score >= {threshold}... (sin resultados)")

    if not top_content:
        print("⚠️ No se encontraron resultados relevantes con ningún umbral.")
        return []
    return top_content

    # thresholds = [2, 1.5, 1, 0.5, 0]

    # # 4. Recopilar y devolver el contenido de los resultados
    # top_content = [content for content in results if content['@search.reranker_score'] > 2]  # Filtro de confianza
    # # top_content = [content for content in results]
    
    # if not top_content:
    #     print("⚠️ No se encontraron resultados con un puntaje de reclasificación suficientemente alto.")
    #     return []
        
    # # print(f"📚 Se encontraron {len(top_content)} fragmentos de texto relevantes.")
    # return top_content

pregunta_usuario = "¿Qué se determinó sobre la participación de la guerrilla del M-19 en los eventos de la masacre de El Salado?"

normalized_query = normalize_text(pregunta_usuario)

# #normalized_query = "cuantos años de prision se le impuso a jorge barney veloza garcia"

retrieved_context = hybrid_search(normalized_query, "index_sentencias")
retrieved_context
```

```python
import re

def clean_highlights(text):
    stopwords = {
        'el', 'la', 'los', 'las', 'un', 'una', 'unos', 'unas',
        'de', 'del', 'al', 'a', 'en', 'por', 'para', 'con', 'sin',
        'y', 'o', 'u', 'que', 'como', 'es', 'su', 'sus', 'se', 'lo', 'le', 'les',
        'mi', 'mis', 'tu', 'tus', 'su', 'sus', 'nuestro', 'nuestra', 'nuestros', 'nuestras'
    }
    # Busca todas las ocurrencias de <em>palabra</em>
    def replace_em(match):
        word = match.group(1)
        if word.lower() in stopwords:
            return word  # Quita el <em>, deja solo la palabra
        else:
            return match.group(0)  # Deja el <em> intacto

    # Expresión regular para buscar <em>...</em>
    cleaned = re.sub(r'<em>(.*?)</em>', replace_em, text)
    return cleaned
# Ejemplo de uso:
resaltado = "Este es un <em>ejemplo</em> de <em>la</em> función para <em>resaltado</em> de palabras <em>de</em> interés."
limpio = clean_highlights(resaltado)
print(limpio)
```
